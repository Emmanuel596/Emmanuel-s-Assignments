Import pandas library and read data

```python
import pandas as pd
Raw_dataset = pd.read_csv(r"C:\Users\His Grace\Desktop\Yuvainternship\archive (5)\customer_master.csv")
print(Raw_dataset)
```

Checking data integrity

```python
print(Raw_dataset.columns)
print(Raw_dataset.isna().sum())
print(Raw_dataset.duplicated().sum())
```

Perform Descriptive analysis

```python
print(Raw_dataset.describe())
```

```python
Man = Raw_dataset.gender.value_counts()
New_parameter = Raw_dataset.customer_country.value_counts()
print(New_parameter)
print(Man)
```

```python
import matplotlib.pyplot as plt
x = New_parameter.index
y = New_parameter.values
colors = ['green', 'skyblue', 'blue', 'yellow', 'pink', 'violet', 'red']
plt.bar(x,y, color = colors , edgecolor = 'black', width=0.7)
plt.xlabel('Country')
plt.ylabel('Country Counts')
plt.title('Comparing Customer Countries')
for i, value in enumerate(y):
    plt.text(i, value +100, str(value), ha = 'center', va = 'bottom', fontweight = 'bold')
plt.grid(axis='y', linestyle = '--', alpha = 0.7)
plt.show()
```

```python
New_data = pd.crosstab(Raw_dataset['customer_country'], Raw_dataset['customer_segment'])
print(New_data)
```
