#Data visuaisation using Python's bar chart

```python
import matplotlib.pyplot as plt
x = Raw_dataset['price_category']
y = Raw_dataset['monthly_maintenance_inr']
plt.bar(x,y, color = 'blue', edgecolor = 'black', width= 0.3)
plt.title ("Average Monthly Maintenance by Price Category")
plt.xlabel ('price_category')
plt.ylabel ('monthly_maintenance_inr')
plt.grid(axis='y', linestyle = '--')
plt.show()
```
